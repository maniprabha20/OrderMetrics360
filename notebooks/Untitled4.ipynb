```python
delivery_summary_df = client.query("""
SELECT
    CASE
        WHEN delivery_delay_days <= 0 THEN 'On Time / Early'
        ELSE 'Delayed'
    END AS delivery_status,
    COUNT(*) AS order_count,
    ROUND(AVG(review_score), 2) AS average_review_score,
    ROUND(AVG(delivery_delay_days), 2) AS average_delay_days
FROM `ordermetrics360.ordermetrics360_marts.mart_delivery_reviews`
WHERE review_score IS NOT NULL
GROUP BY delivery_status
ORDER BY delivery_status
""").to_dataframe()

delivery_summary_df
```

```python
delivery_summary_df.to_csv(
    'data/processed/delivery_summary.csv',
    index=False
)
```
